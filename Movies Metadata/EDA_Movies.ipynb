```python
import ast

def get_collection_name(x):
  if isinstance(x, float):
    return ""
  try:
    dict_data = ast.literal_eval(x)
    return dict_data['name']
  except (ValueError, SyntaxError, KeyError):
    return ""
```

```python
df_clean['belongs_to_collection'] = df_clean['belongs_to_collection'].apply(get_collection_name)
```

```python
def extract_names_from_list(x):
  if isinstance(x, list):
    return x

  # Tangani Missing Values
  if isinstance(x, float):
    return ""

  # Proses ekstraksi kunci 'name'
  try:
    data_list = ast.literal_eval(x)
    return [item['name']for item in data_list]
  except (ValueError, SyntaxError, KeyError, TypeError):
    return ""
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. PARSING: MENGUBAH LIST MENJADI STRING
# ==========================================
# Fungsi untuk menggabungkan elemen list menggunakan spasi
df_clean['genres'] = df_clean['genres'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x))

print("Contoh hasil pembongkaran genres:")
display(df_clean[['title', 'genres']].head())

# ==========================================
# 2. VISUALISASI EDA: TOP 10 GENRE FILM
# ==========================================
# Memisahkan kembali string genre per kata khusus untuk dihitung statistiknya
semua_genre = df_clean['genres'].str.split(' ').explode().reset_index(drop=True)
# Catatan: Hapus string kosong jika ada film tanpa genre
semua_genre = semua_genre[semua_genre != ""]

# Membuat visualisasi
plt.figure(figsize=(12, 6))
sns.countplot(
    y=semua_genre,
    order=semua_genre.value_counts().index[:10],
    hue=semua_genre,
    legend=False,
    palette='magma'
)
plt.title('Top 10 Genre Film Terbanyak di Dataset TMDB', fontsize=14, fontweight='bold')
plt.xlabel('Jumlah Film')
plt.ylabel('Genre')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
